```python
# Missing-value analysis

for name, df in datasets.items():

    print("\n" + "=" * 60)
    print(f"Missing Value Analysis: {name}")
    print("=" * 60)

    missing_values = df.isnull().sum()

    missing_percentage = (
        df.isnull().mean() * 100
    ).round(2)

    result = pd.DataFrame({
        "Missing Count": missing_values,
        "Missing Percentage": missing_percentage
    })

    display(result)
```

```python

# Prepare expected baggage journey details

expected_journeys = (
    baggage_assignment
    .merge(flight, on="flight_id", how="left")
    .sort_values(["bag_id", "segment_sequence"])
    .reset_index(drop=True)
)

# Convert flight timestamps to datetime
expected_journeys["departure_time"] = pd.to_datetime(
    expected_journeys["departure_time"],
    utc=True
)

expected_journeys["arrival_time"] = pd.to_datetime(
    expected_journeys["arrival_time"],
    utc=True
)

# Check whether arrival happens after departure
expected_journeys["valid_flight_schedule"] = (
    expected_journeys["arrival_time"]
    > expected_journeys["departure_time"]
)

print("Total assigned flight segments:", len(expected_journeys))

print(
    "Invalid flight schedules:",
    (~expected_journeys["valid_flight_schedule"]).sum()
)

display(
    expected_journeys[
        ~expected_journeys["valid_flight_schedule"]
    ]
)
```

```python

# Connecting flight chronology validation

# Get the first and second flight segments for each bag
first_flight = (
    expected_journeys[
        expected_journeys["segment_sequence"] == 1
    ][["bag_id", "source", "destination", "departure_time", "arrival_time"]]
    .rename(columns={
        "source": "first_source",
        "destination": "first_destination",
        "departure_time": "first_departure",
        "arrival_time": "first_arrival"
    })
)

second_flight = (
    expected_journeys[
        expected_journeys["segment_sequence"] == 2
    ][["bag_id", "source", "destination", "departure_time", "arrival_time"]]
    .rename(columns={
        "source": "second_source",
        "destination": "second_destination",
        "departure_time": "second_departure",
        "arrival_time": "second_arrival"
    })
)

# Join both flight segments using bag_id
connections = first_flight.merge(
    second_flight,
    on="bag_id",
    how="inner"
)

# Calculate connection time in minutes
connections["connection_gap_minutes"] = (
    connections["second_departure"] -
    connections["first_arrival"]
).dt.total_seconds() / 60

# Validate airport continuity and minimum connection time
connections["airport_continuity"] = (
    connections["first_destination"] ==
    connections["second_source"]
)

connections["valid_connection_time"] = (
    connections["connection_gap_minutes"] >= 60
)

connections["valid_connection"] = (
    connections["airport_continuity"] &
    connections["valid_connection_time"]
)

print("Total connecting journeys:", len(connections))
print("Invalid airport continuity:",
      (~connections["airport_continuity"]).sum())
print("Invalid connection time:",
      (~connections["valid_connection_time"]).sum())
print("Total invalid connections:",
      (~connections["valid_connection"]).sum())

display(
    connections[
        ~connections["valid_connection"]
    ].head(10)
)
```

```python

# Analyze invalid connection time distribution

invalid_connections = connections[
    ~connections["valid_connection_time"]
].copy()

print("Invalid connection gap statistics:")
display(
    invalid_connections["connection_gap_minutes"].describe()
)

print("\nConnection gap frequency:")
display(
    invalid_connections["connection_gap_minutes"]
    .value_counts()
    .sort_index()
    .to_frame(name="Number of Bags")
)

print("\nInvalid connection examples:")
display(
    invalid_connections[
        [
            "bag_id",
            "first_arrival",
            "second_departure",
            "connection_gap_minutes"
        ]
    ].head(10)
)
```

```python

# Prepare expected airports for each baggage journey

expected_airports = (
    expected_journeys[
        ["bag_id", "segment_sequence", "source", "destination"]
    ]
)

# Join scan events with expected baggage flight segments
scan_route_check = baggage_scan_event.merge(
    expected_airports,
    on="bag_id",
    how="left"
)

# Check whether the scan airport matches either endpoint
scan_route_check["airport_matches"] = (
    (scan_route_check["airport_code"] == scan_route_check["source"]) |
    (scan_route_check["airport_code"] == scan_route_check["destination"])
)

print("Total joined records:", len(scan_route_check))

print(
    "Matching route records:",
    scan_route_check["airport_matches"].sum()
)

print(
    "Non-matching route records:",
    (~scan_route_check["airport_matches"]).sum()
)
```

```python

# Collect all expected airports for each baggage journey

expected_airports_per_bag = (
    expected_airports
    .groupby("bag_id")
    .apply(
        lambda group: set(group["source"]) | set(group["destination"])
    )
    .reset_index(name="expected_airports")
)

# Join each original scan event with its baggage's expected airports
scan_route_validation = baggage_scan_event.merge(
    expected_airports_per_bag,
    on="bag_id",
    how="left"
)

# Check whether the scanned airport belongs to the expected journey
scan_route_validation["airport_matches"] = (
    scan_route_validation.apply(
        lambda row: row["airport_code"] in row["expected_airports"],
        axis=1
    )
)

print("Original scan events:", len(scan_route_validation))
print(
    "Matching scan events:",
    scan_route_validation["airport_matches"].sum()
)
print(
    "Non-matching scan events:",
    (~scan_route_validation["airport_matches"]).sum()
)
```

```python

# Sort scan events by baggage ID and event timestamp

scan_events_sorted = baggage_scan_event.copy()

scan_events_sorted["event_time"] = pd.to_datetime(
    scan_events_sorted["event_time"],
    utc=True
)

scan_events_sorted = scan_events_sorted.sort_values(
    ["bag_id", "event_time"]
)

# Display event sequences for one direct and one connecting journey

for bag_id in ["BAG00001", "BAG00501"]:
    print(f"\nEvent sequence for {bag_id}:")

    display(
        scan_events_sorted.loc[
            scan_events_sorted["bag_id"] == bag_id,
            ["event_id", "event_type", "airport_code", "event_time"]
        ]
    )
```

```python

# Check whether event timestamps are strictly increasing
# according to the original event record order for each baggage.

event_order_check = baggage_scan_event.copy()

event_order_check["event_time"] = pd.to_datetime(
    event_order_check["event_time"],
    utc=True
)

# Preserve the original order of records using event_id
event_order_check = event_order_check.sort_values(
    ["bag_id", "event_id"]
)

event_order_check["previous_event_time"] = (
    event_order_check.groupby("bag_id")["event_time"].shift(1)
)

event_order_check["timestamp_increases"] = (
    event_order_check["previous_event_time"].isna() |
    (
        event_order_check["event_time"] >
        event_order_check["previous_event_time"]
    )
)

# Summarize baggage with timestamp-order violations
bags_with_violations = (
    event_order_check.groupby("bag_id")["timestamp_increases"]
    .all()
)

print("Total baggage:", len(bags_with_violations))
print(
    "Bags with timestamp-order violations:",
    (~bags_with_violations).sum()
)
print(
    "Bags with increasing timestamps:",
    bags_with_violations.sum()
)
```

```python

# Identify baggage IDs with timestamp-order violations
timestamp_violation_bags = set(
    bags_with_violations[~bags_with_violations].index
)

# Identify baggage IDs with invalid connecting-flight timing
invalid_connection_bags = set(
    invalid_connections["bag_id"]
)

# Compare both groups
common_bags = timestamp_violation_bags & invalid_connection_bags

print("Bags with timestamp violations:", len(timestamp_violation_bags))
print("Bags with invalid connections:", len(invalid_connection_bags))
print("Bags present in both groups:", len(common_bags))

print(
    "Timestamp violations only:",
    len(timestamp_violation_bags - invalid_connection_bags)
)

print(
    "Invalid connections only:",
    len(invalid_connection_bags - timestamp_violation_bags)
)
```

```python

# Build the event-type sequence using event_id order

event_sequences = (
    event_order_check
    .sort_values(["bag_id", "event_id"])
    .groupby("bag_id")["event_type"]
    .agg(lambda events: " -> ".join(events))
    .reset_index(name="event_sequence")
)

# Count how many bags share each sequence
sequence_summary = (
    event_sequences
    .groupby("event_sequence")
    .size()
    .reset_index(name="number_of_bags")
    .sort_values("number_of_bags", ascending=False)
)

print("Number of unique event-sequence patterns:",
      len(sequence_summary))

display(sequence_summary)
```

```python

# Create a working copy without modifying raw data

scan_events_sorted = baggage_scan_event.copy()

# Convert event_time into UTC datetime format
scan_events_sorted["event_time"] = pd.to_datetime(
    scan_events_sorted["event_time"],
    utc=True
)

# Verify the conversion
print("Event timestamp data type:")
print(scan_events_sorted["event_time"].dtype)

print("\nSample event timestamps:")
display(
    scan_events_sorted[
        ["event_id", "bag_id", "event_type", "event_time"]
    ].head()
)
```

```python

# Sort events by baggage ID and event ID

event_order_check = scan_events_sorted.sort_values(
    ["bag_id", "event_id"]
).copy()

# Get the previous event timestamp for each baggage
event_order_check["previous_event_time"] = (
    event_order_check.groupby("bag_id")["event_time"].shift(1)
)

# Check whether the current event timestamp is greater
# than the previous event timestamp
event_order_check["timestamp_increases"] = (
    event_order_check["previous_event_time"].isna() |
    (
        event_order_check["event_time"] >
        event_order_check["previous_event_time"]
    )
)

# Count baggage records with timestamp violations
bags_with_violations = (
    event_order_check.groupby("bag_id")["timestamp_increases"].all()
)

print("Total baggage:", len(bags_with_violations))
print(
    "Bags with timestamp violations:",
    (~bags_with_violations).sum()
)
print(
    "Bags with increasing timestamps:",
    bags_with_violations.sum()
)
```

```python

# Filter events where timestamp does not increase

timestamp_violations = event_order_check[
    ~event_order_check["timestamp_increases"]
].copy()

print("Total timestamp violations:", len(timestamp_violations))

print("\nSample violating events:")
display(
    timestamp_violations[
        [
            "bag_id",
            "event_id",
            "event_type",
            "previous_event_time",
            "event_time"
        ]
    ].head(15)
)
```

```python

# Count timestamp violations by event type

violation_event_summary = (
    timestamp_violations["event_type"]
    .value_counts()
    .reset_index()
)

violation_event_summary.columns = [
    "Event Type",
    "Violation Count"
]

display(violation_event_summary)
```

```python

# Count timestamp violations by event type and airport

violation_location_summary = (
    timestamp_violations
    .groupby(["event_type", "airport_code"])
    .size()
    .reset_index(name="Violation Count")
    .sort_values("Violation Count", ascending=False)
)

display(violation_location_summary)
```

```python

# Reconstruct the expected flight segments

journey_segments = (
    baggage_assignment
    .merge(flight, on="flight_id", how="left")
    .sort_values(["bag_id", "segment_sequence"])
)

# Identify the first and second flight for each connecting bag
first_segments = (
    journey_segments[journey_segments["segment_sequence"] == 1]
    [["bag_id", "destination"]]
    .rename(columns={"destination": "connection_airport"})
)

second_segments = (
    journey_segments[journey_segments["segment_sequence"] == 2]
    [["bag_id", "source"]]
    .rename(columns={"source": "second_flight_source"})
)

connection_airports = first_segments.merge(
    second_segments,
    on="bag_id",
    how="inner"
)

# Compare the violating LOADED event airport with the expected connection
violation_airport_check = timestamp_violations[
    ["bag_id", "event_type", "airport_code"]
].merge(
    connection_airports,
    on="bag_id",
    how="left"
)

violation_airport_check["is_connection_airport"] = (
    (violation_airport_check["airport_code"] ==
     violation_airport_check["connection_airport"]) &
    (violation_airport_check["airport_code"] ==
     violation_airport_check["second_flight_source"])
)

print("Total violations:", len(violation_airport_check))
print(
    "Violations at expected connection airport:",
    violation_airport_check["is_connection_airport"].sum()
)
print(
    "Violations at unexpected airport:",
    (~violation_airport_check["is_connection_airport"]).sum()
)

display(violation_airport_check.head(10))
```

```python

# Get the second flight details for each connecting baggage

second_flight_details = journey_segments[
    journey_segments["segment_sequence"] == 2
][
    ["bag_id", "flight_number", "source", "destination", "departure_time"]
].copy()

second_flight_details["departure_time"] = pd.to_datetime(
    second_flight_details["departure_time"],
    utc=True
)

# Get the affected LOADED events
loaded_violations = timestamp_violations[
    timestamp_violations["event_type"] == "LOADED"
][
    ["bag_id", "airport_code", "event_time"]
].copy()

# Compare actual loading time with scheduled departure
loading_schedule_check = loaded_violations.merge(
    second_flight_details,
    on="bag_id",
    how="left"
)

loading_schedule_check["loading_to_departure_minutes"] = (
    loading_schedule_check["departure_time"] -
    loading_schedule_check["event_time"]
).dt.total_seconds() / 60

print("Total records:", len(loading_schedule_check))

display(
    loading_schedule_check[
        [
            "bag_id",
            "airport_code",
            "flight_number",
            "event_time",
            "departure_time",
            "loading_to_departure_minutes"
        ]
    ].head(10)
)
```

```python
import hashlib

def calculate_hash(file_path):
    sha256 = hashlib.sha256()

    with open(file_path, "rb") as file:
        for chunk in iter(lambda: file.read(4096), b""):
            sha256.update(chunk)

    return sha256.hexdigest()


for file_name in [
    "baggage.csv",
    "flight.csv",
    "baggage_assignment.csv",
    "baggage_scan_event.csv"
]:
    raw_file = raw_path / file_name
    working_file = working_path / file_name

    raw_hash = calculate_hash(raw_file)
    working_hash = calculate_hash(working_file)

    print(f"\nDataset: {file_name}")
    print("Raw and working contents identical:", raw_hash == working_hash)
```

```python
# Join assignments with flight schedules
assignment_schedule_check = assignment_work.merge(
    flight_work[
        ["flight_id", "departure_time", "arrival_time"]
    ],
    on="flight_id",
    how="left",
    validate="many_to_one"
)

# Check assignment time against flight departure
assignment_schedule_check["assigned_before_departure"] = (
    assignment_schedule_check["assigned_at"] <
    assignment_schedule_check["departure_time"]
)

print("Total assignments:", len(assignment_schedule_check))

print(
    "Assigned before departure:",
    assignment_schedule_check["assigned_before_departure"].sum()
)

print(
    "Assigned at or after departure:",
    (~assignment_schedule_check["assigned_before_departure"]).sum()
)

display(
    assignment_schedule_check[
        ~assignment_schedule_check["assigned_before_departure"]
    ]
)
```

```python
# Validate event occurrence and receipt chronology

scan_event_work["receipt_after_event"] = (
    scan_event_work["received_at"] >=
    scan_event_work["event_time"]
)

print("Total scan events:", len(scan_event_work))

print(
    "Receipt at or after event:",
    scan_event_work["receipt_after_event"].sum()
)

print(
    "Receipt before event:",
    (~scan_event_work["receipt_after_event"]).sum()
)

display(
    scan_event_work[
        ~scan_event_work["receipt_after_event"]
    ]
)
```

```python
# Create final cleaned DataFrames without temporary validation columns

baggage_clean = baggage_work.copy()

flight_clean = flight_work.drop(
    columns=["schedule_valid"],
    errors="ignore"
).copy()

assignment_clean = assignment_work.copy()

scan_event_clean = scan_event_work.drop(
    columns=["receipt_after_event"],
    errors="ignore"
).copy()

# Organize cleaned datasets
cleaned_datasets = {
    "Baggage": baggage_clean,
    "Flight": flight_clean,
    "Baggage Assignment": assignment_clean,
    "Baggage Scan Event": scan_event_clean
}

# Verify dimensions
for name, df in cleaned_datasets.items():
    print(f"{name}: {df.shape}")
```
