```python
import pandas as pd
import numpy as np
from pathlib import Path

project_path = Path.cwd()
data_path = project_path / "data"
processed_path = data_path / "processed"

print("Project path:", project_path)
print("Processed folder exists:", processed_path.exists())
```

```python
input_file = processed_path / "final_phase11_advanced_investigation.csv"

exceptions_source = pd.read_csv(input_file)

print("Dataset loaded successfully!")
print("Shape:", exceptions_source.shape)
print("\nColumns:")
print(exceptions_source.columns.tolist())

display(exceptions_source.head())
```

```python
print("Original Investigation Status:")
display(
    exceptions_source["investigation_status"].value_counts()
)

print("\nAdvanced Detection Status:")
display(
    exceptions_source["advanced_detection_status"].value_counts()
)

print("\nAdvanced Severity:")
display(
    exceptions_source["advanced_severity"].value_counts()
)

print("\nDetection Flags:")
flag_columns = [
    "missing_segment_event_detected",
    "handoff_timing_inconsistency_detected",
    "airport_consistency_issue_detected",
    "event_duration_pattern_detected"
]

for column in flag_columns:
    print(f"\n{column}")
    print(exceptions_source[column].value_counts())
```

```python
exception_records = []

for _, row in exceptions_source.iterrows():

    # Exception 1: Missing segment event
    if row["missing_segment_event_detected"]:

        exception_records.append({
            "bag_id": row["bag_id"],
            "exception_type": "MISSING_SEGMENT_EVENT",
            "exception_description": (
                f"Missing event: {row['missing_segment_events']}"
            ),
            "severity": "MEDIUM",
            "priority": 2
        })

    # Exception 2: Handoff timing inconsistency
    if row["handoff_timing_inconsistency_detected"]:

        exception_records.append({
            "bag_id": row["bag_id"],
            "exception_type": "HANDOFF_TIMING_INCONSISTENCY",
            "exception_description": (
                "Second flight loading occurred before "
                "the baggage transfer event."
            ),
            "severity": "HIGH",
            "priority": 1
        })

exceptions_df = pd.DataFrame(exception_records)

print("Exception generation completed!")
print("Total exceptions:", len(exceptions_df))
print("Unique baggage IDs:", exceptions_df["bag_id"].nunique())

print("\nException type distribution:")
display(exceptions_df["exception_type"].value_counts())

print("\nSeverity distribution:")
display(exceptions_df["severity"].value_counts())

display(exceptions_df.head(10))
```

```python
# Generate unique exception IDs

exceptions_df.insert(
    0,
    "exception_id",
    [f"EXC{i:06d}" for i in range(1, len(exceptions_df) + 1)]
)

print("Exception IDs generated!")

print("Total exception IDs:", exceptions_df["exception_id"].nunique())

display(exceptions_df.head(10))
```

```python
# Add exception lifecycle management columns

exceptions_df["status"] = "OPEN"
exceptions_df["assigned_to"] = None
exceptions_df["created_at"] = pd.Timestamp.now(tz="UTC")
exceptions_df["updated_at"] = pd.Timestamp.now(tz="UTC")
exceptions_df["resolved_at"] = pd.NaT
exceptions_df["resolution_notes"] = None

print("Exception lifecycle columns added!")

print("\nColumns:")
print(exceptions_df.columns.tolist())

print("\nException status distribution:")
display(exceptions_df["status"].value_counts())

display(exceptions_df.head())
```

```python
def update_exception_status(
    exception_id,
    new_status,
    assigned_to=None,
    resolution_notes=None
):
    valid_statuses = ["OPEN", "IN_PROGRESS", "RESOLVED"]

    if new_status not in valid_statuses:
        raise ValueError("Invalid exception status!")

    matching = exceptions_df.index[
        exceptions_df["exception_id"] == exception_id
    ]

    if len(matching) == 0:
        raise ValueError("Exception ID not found!")

    idx = matching[0]
    current_status = exceptions_df.at[idx, "status"]

    allowed_transitions = {
        "OPEN": ["IN_PROGRESS"],
        "IN_PROGRESS": ["OPEN", "RESOLVED"],
        "RESOLVED": []
    }

    if new_status not in allowed_transitions[current_status]:
        raise ValueError(
            f"Cannot change status from {current_status} to {new_status}"
        )

    if new_status == "IN_PROGRESS" and not assigned_to:
        raise ValueError("Investigator assignment is required!")

    if new_status == "RESOLVED" and not resolution_notes:
        raise ValueError("Resolution notes are required!")

    now = pd.Timestamp.now(tz="UTC")

    exceptions_df.at[idx, "status"] = new_status
    exceptions_df.at[idx, "updated_at"] = now

    if new_status == "IN_PROGRESS":
        exceptions_df.at[idx, "assigned_to"] = assigned_to

    if new_status == "RESOLVED":
        exceptions_df.at[idx, "resolved_at"] = now
        exceptions_df.at[idx, "resolution_notes"] = resolution_notes

    return exceptions_df.loc[idx]
```

```python
print("Exception Lifecycle Summary")

display(
    exceptions_df["status"].value_counts().rename_axis(
        "status"
    ).reset_index(name="total_exceptions")
)

print("\nResolved exception details:")

display(
    exceptions_df[
        exceptions_df["status"] == "RESOLVED"
    ][
        [
            "exception_id",
            "bag_id",
            "severity",
            "assigned_to",
            "resolved_at",
            "resolution_notes"
        ]
    ]
)
```

```python
print("EXCEPTION DATA QUALITY VALIDATION")

print("\n1. Duplicate Exception IDs:")
print(exceptions_df["exception_id"].duplicated().sum())

print("\n2. Missing Exception IDs:")
print(exceptions_df["exception_id"].isna().sum())

print("\n3. Missing Baggage IDs:")
print(exceptions_df["bag_id"].isna().sum())

print("\n4. Invalid Status Values:")
valid_statuses = ["OPEN", "IN_PROGRESS", "RESOLVED"]
print(
    (~exceptions_df["status"].isin(valid_statuses)).sum()
)

print("\n5. Resolved Exceptions Without Resolution Timestamp:")
print(
    (
        (exceptions_df["status"] == "RESOLVED")
        & exceptions_df["resolved_at"].isna()
    ).sum()
)

print("\n6. Resolved Exceptions Without Resolution Notes:")
print(
    (
        (exceptions_df["status"] == "RESOLVED")
        & exceptions_df["resolution_notes"].isna()
    ).sum()
)
```

```python
exception_summary = (
    exceptions_df
    .groupby(
        ["exception_type", "severity", "status"],
        dropna=False
    )
    .agg(
        total_exceptions=("exception_id", "count"),
        unique_baggage=("bag_id", "nunique")
    )
    .reset_index()
)

print("EXCEPTION MANAGEMENT SUMMARY")

display(exception_summary)

print("\nOverall status:")
display(exceptions_df["status"].value_counts())

print("\nSeverity distribution:")
display(exceptions_df["severity"].value_counts())
```
